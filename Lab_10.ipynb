```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture
from sklearn.datasets import make_blobs
```

```python
X, y_true = make_blobs(n_samples=300, centers=3, cluster_std=0.8, random_state=42)
```

```python
gmm = GaussianMixture(n_components=3, random_state=42)
gmm.fit(X)
```

```python
labels = gmm.predict(X)
```

```python
print("Converged:", gmm.converged_)
print("Number of iterations run:", gmm.n_iter_)
print("Means of each Gaussian component:\n", gmm.means_)
print("Weights of each component:\n", gmm.weights_)
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap="viridis", s=30, alpha=0.7)
plt.scatter(gmm.means_[:, 0], gmm.means_[:, 1], c="red", marker="X", s=200, label="Cluster Centers")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("EM Algorithm (Gaussian Mixture Model) Clustering")
plt.legend()
plt.show()
```
